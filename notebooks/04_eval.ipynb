# 04 — Kết quả đánh giá (M7)
Đọc `results/*/metrics.json` và `results/ablation.csv`.

```python
import json, pathlib
import pandas as pd  # nếu chưa có: pip install pandas
```

```python
rows = []
for p in sorted(pathlib.Path("../results").glob("*/metrics.json")):
    m = json.loads(p.read_text(encoding="utf-8"))
    pm = m[m["primary"]]
    rows.append({"run": p.parent.name, "eer": pm.get("eer"), "ci95": pm.get("eer_ci95"),
                 "frr@far": pm.get("frr_at_far"), "fta": m["fta"]["fta"],
                 "thr@far": pm.get("threshold_at_far")})
pd.DataFrame(rows)
```

```python
p = pathlib.Path("../results/ablation.csv")
pd.read_csv(p) if p.exists() else "chưa chạy ablation"
```
