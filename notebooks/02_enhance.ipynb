Duyệt `04_flat`, `05_enhanced`; so sánh `enhance.invert` và `enhance.channel`. Đổi config, không sửa code.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import cv2, numpy as np, matplotlib.pyplot as plt
import fpsys
from fpsys.pipeline import run
cfg = fpsys.load_config()
```

```python
def show(steps, names, cols=3, size=4):
    rows = (len(names) + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(cols * size, rows * size))
    for ax in np.ravel(axes): ax.axis("off")
    for ax, n in zip(np.ravel(axes), names):
        if n in steps:
            im = steps[n]
            ax.imshow(im if im.ndim == 2 else im[..., ::-1], cmap="gray"); ax.set_title(n)
    plt.tight_layout()
```

```python
p = sorted(pathlib.Path("../data/own").glob("*.jpg"))[0]
img = fpsys.load_image(p)
for ov in ([], ["enhance.invert=false"], ["enhance.channel=green"]):
    tr = run(img, fpsys.load_config(overrides=ov), collect=True)
    print(ov, tr.error)
    show(tr.steps, ["03_norm", "04_flat", "05_enhanced", "06_orientation", "07_binary", "08_skeleton"])
    plt.show()
```
