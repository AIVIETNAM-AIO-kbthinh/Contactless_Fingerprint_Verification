# 03 — Minutiae (M4)
Đếm minutiae, xem phân bố chất lượng và vị trí.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import cv2, numpy as np, matplotlib.pyplot as plt
import fpsys
from fpsys.pipeline import run
cfg = fpsys.load_config()
```

```python
def show(steps, names, cols=3, size=4):
    rows = (len(names) + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(cols * size, rows * size))
    for ax in np.ravel(axes): ax.axis("off")
    for ax, n in zip(np.ravel(axes), names):
        if n in steps:
            im = steps[n]
            ax.imshow(im if im.ndim == 2 else im[..., ::-1], cmap="gray"); ax.set_title(n)
    plt.tight_layout()
```

```python
paths = sorted(pathlib.Path("../data/own").glob("*.jpg"))[:10]
for p in paths:
    tr = run(fpsys.load_image(p), cfg, collect=True)
    n = len(tr.template.minutiae) if tr.template else 0
    print(p.name, n, tr.error)
    show(tr.steps, ["08_skeleton", "09_minutiae"], cols=2, size=6)
    plt.show()
```
